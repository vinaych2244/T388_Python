```python
n = 15
for i in range (n,0,-1):
    print("* " *i)
for i in range (1,n+1):
    print(" " *((2*n)-(2*i)),"* "*i)
for i in range (1,n+1):
    print(" "*(n-i),"* "*i)
for i in range(n, 0, -1):
    print(" " * (n - i) + "* " * i)

# Lower half (normal pyramid)
for i in range(2, n + 1):
    print(" " * (n - i) + "* " * i)
    
```

```python
import numpy as np
def speed(u,a,s):
    
    v_square=(u**2)+(2*a*s)
    speed = np.sqrt(v_square)
    print("speed is",round(speed,2),"m/s")
initial_vel= int(input("Enter INitial velocity"))
acceleration= int(input("Enter acceleration of the car"))
distance= int(input("Enter the distance between source and destination"))
speed(initial_vel,acceleration,distance)
    
```

```python
"""1. Find the Second Largest Number
	• Task: Find the second largest unique number in a list without using sort() or max().
	• Example: [10, 20, 4, 45, 99, 99] → 45
"""
```

```python
list1=[10, 20, 4, 45, 99, 99] 
list2=[i for i in list1 if i== i]
```

```python
nums = [10, 20, 4, 45, 99, 99]

largest = float('-inf')
second_largest = float('-inf')

for num in nums:
    if num > largest:
        second_largest = largest
        largest = num
    elif second_largest < num < largest:
        second_largest = num

if second_largest == float('-inf'):
    print("No second largest unique number found.")
else:
    print("Second largest:", second_largest)
```

```python
"""2. Count Even and Odd Numbers
	• Task: Given a list of integers, return a count of how many are even and how many are odd.
	• Example: [1, 2, 3, 4, 5] → Even: 2, Odd: 3
"""
even = 0
Odd =0
list1=[1, 2, 3, 4, 5]
for i in list1:
    if i%2==0:
        even+=1
    else:
        Odd+=1
print(f"even:{even},odd:{Odd}")
```
